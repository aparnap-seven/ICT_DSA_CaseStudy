1. Bank Account Manager
a. Simulate a bank account with deposit, withdrawal, and balance
enquiry operations
b. Use variables and data types to store account information
c. Implement conditional statements to check for sufficient balance
before withdrawal
d. Use functions to perform deposit, withdrawal, and balance inquiry
operations.

```python
#Bank Account Details
class BankAccount:
   def __init__(self, owner, accno, balance):
    self.owner = owner
    self.accno = accno
    self.balance = balance


#Function to Deposit
   def deposit(self, amount):
    self.balance = self.balance + amount
    print("Your current balance is: ", self.balance)


#Function to Withdraw
   def withdraw(self, amount):
    if amount> self.balance:
      print("Not sufficient balance")
    else:
      self.balance = self.balance - amount
    print("Amount Withdrawn: ", self.balance)


#Function for balance withdrawal
   def checkbalance(self):
    print(f"Account Name: {self.owner} ")
    print(f"Account Number: {self.accno} ")
    print(f"Your Current balance: {self.balance} ")

#To check customers
customer1 = BankAccount("Aparna", 20641, 5000)
customer2 = BankAccount("Anjali", 20642, 3000)
customer3 = BankAccount("Adithya", 20643, 9000)

customer1.deposit(500)
customer1.withdraw(1000)
customer1.checkbalance()
```

2. Weather Data Analyzer
a. Analyze weather data (temperature, humidity, etc.) for a given region
b. Use variables and data types to store weather data
c. Implement conditional statements to determine weather conditions
(sunny, rainy, etc.)
d. Use functions to calculate average temperature and humidity, and print
results


```python
#Weather data
temperature = [34, 36, 32, 28]
humidity = [60, 65, 70, 75]

#To calculate average temparature and humidity
def avg_temperature(temperature):
  return sum(temperature)/len(temperature)

def avg_humidity(humidity):
  return sum(humidity)/len(humidity)

#To determine weather condition
def weather_condition(avg_temperature, avg_humidity):
  if avg_temperature > 30 and avg_humidity < 60:
        return "Sunny"
  elif avg_humidity > 70:
        return "Rainy"
  else:
        return "Cloudy"

avg_temp = avg_temperature(temperature)
avg_hum = avg_humidity(humidity)

#To display Output
print(f"Average temperature: {avg_temp}")
print(f"Average humidity: {avg_hum}")
print(f"Weather condition: {weather_condition(avg_temp, avg_hum)}")
```
