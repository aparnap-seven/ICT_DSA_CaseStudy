3. Handling Missing Data:
Identify if there are any missing values in the dataset. Replace missing values in the
horsepower column with the column's median.


```python
#To check values in Horsepower and find '?'
print(df['horsepower'].unique())
#To convert the'?' to numeric form
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
median_hp = df['horsepower'].median()
df['horsepower'] = df['horsepower'].fillna(median_hp)
print(df.isna().sum())
```

```python
#To verify that all '?' are replaced now
print((df['horsepower'] == '?').sum())
```

10. Finding Anomalies -
Identify potential outliers in the mpg column using the Interquartile Range (IQR)
method. Specifically:
● Calculate the IQR for mpg. ● Define outliers as values less than Q1 - 1.5 * IQR or greater than Q3 +
1.5 * IQR.
● Create a DataFrame of cars classified as outliers, displaying car_name, mpg,
and model_year.

```python
#To calculate Q1, Q3 and IQR
Q1 = df['mpg'].quantile(0.25)
Q3 = df['mpg'].quantile(0.75)
IQR = Q3-Q1

#To define outlier condition
outliers = df[(df['mpg'] < (Q1 - 1.5 * IQR)) |
              (df['mpg'] > (Q3 + 1.5 * IQR))]
cars = outliers[['car name', 'mpg', 'model year']]
print(cars)
```

# NumPy Problems

1. Basic Array Operations :
Convert the mpg column into a NumPy array and calculate:
○ The mean, median, and standard deviation of mpg.
○ The number of cars with mpg greater than 25.

```python
import numpy as np
mpg_array = df['mpg'].values

#To find mean, median and standard deviation
mean_mpg = np.mean(mpg_array)
median_mpg = np.median(mpg_array)
std_mpg = np.std(mpg_array)

print("Mean:", mean_mpg)
print("Median:", median_mpg)
print("Standard Deviation:", std_mpg)

# Count cars with mpg > 25
count = np.sum(mpg_array > 25)

print("Number of cars with mpg > 25:", count)
```

2. Filtering :
Using NumPy, filter all cars with more than 6 cylinders.
Return the corresponding car_name as a list.

```python
#To filter all cars with more than 6 cylinders
cylinders = df['cylinders'].values
car_names = df['car name'].values

# Filter condition
filtered_names = car_names[cylinders > 6]

# Convert to list
result = filtered_names.tolist()

print(result)
```

3. Statistical Analysis :
Compute the 25th, 50th, and 75th percentiles of the weight column using NumPy.

```python
#Converting weight column to NumPy array
weight_array = df['weight'].values
#To calculate percentiles
q1 = np.percentile(weight_array, 25)
q2 = np.percentile(weight_array, 50)
q3 = np.percentile(weight_array, 75)

print("25th percentile (Q1):", q1)
print("50th percentile (Median):", q2)
print("75th percentile (Q3):", q3)
```

4. Array Manipulation :
Convert the acceleration column into a NumPy array and normalize its values
(scale between 0 and 1).

```python
#Convert acceleration to NumPy array
acc_array = df['acceleration'].values
# Find min and max
min_val = np.min(acc_array)
max_val = np.max(acc_array)

# Normalize
normalized = (acc_array - min_val) / (max_val - min_val)
#display first 10 values
print(normalized[:10])
```

5. Broadcasting :
Increase all horsepower values by 10% and store the updated values in a new
NumPy array. Handle missing data (if any) by replacing it with the mean of the
column before applying the increase.

```python
# Convert 'horsepower' to numeric
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')

# Replace missing values with mean
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mean())

# Convert to NumPy array
hp_array = df['horsepower'].values

# Increase horsepower by 10%
new_hp = hp_array * 1.10

print(new_hp[:10])
```

6. Boolean Indexing :
Find the average displacement of cars with an origin of 2 (Europe) using NumPy
indexing.

```python
# Convert required columns to NumPy arrays
displacement = df['displacement'].values
origin = df['origin'].values
# Apply condition (origin == 2)
filtered_dp = displacement[origin == 2]

# Calculate average
avg_displacement = np.mean(filtered_dp)

print("Average displacement (origin = 2):", avg_displacement)
```

7. Matrix Operations :
Create a 2D NumPy array containing the columns mpg, horsepower, and weight.
Compute the dot product of this matrix with a given vector [1, 0.5, -0.2].

```python
# Extract required columns as NumPy arrays
mpg = df['mpg'].values
horsepower = df['horsepower'].values
weight = df['weight'].values
# Create 2D array (matrix)
matrix = np.column_stack((mpg, horsepower, weight))

# Define vector
vector = np.array([1, 0.5, -0.2])

# Compute dot product
result = np.dot(matrix, vector)

print(result[:10])
```

8. Sorting :
Use NumPy to sort the cars by model_year in descending order and display the first
f
ive car names.

```python
# Convert required columns to NumPy arrays
model_year = df['model year'].values
car_names = df['car name'].values

# Get sorted indices (descending order)
sorted = np.argsort(model_year)[::-1]

# Get top 5 car names
top_5_cars = car_names[sorted][:5]

print(top_5_cars)
```

9. Correlation :
Compute the Pearson correlation coefficient between mpg and weight using
NumPy.

```python
# Convert columns to NumPy arrays
mpg = df['mpg'].values
weight = df['weight'].values

# Compute correlation matrix
corr_matrix = np.corrcoef(mpg, weight)

# Extract correlation value
correlation = corr_matrix[0, 1]

print("Pearson correlation:", correlation)
```

10. Conditional Aggregates :
Calculate the mean mpg for cars grouped by the number of cylinders using NumPy
techniques.

```python
#Convert columns to NumPy arrays
mpg = df['mpg'].values
cylinders = df['cylinders'].values

# Get unique cylinder values
unique_cyl = np.unique(cylinders)

# Calculate mean mpg for each group
for cyl in unique_cyl:
    mean_mpg = np.mean(mpg[cylinders == cyl])
    print(f"Cylinders: {cyl}, Mean MPG: {mean_mpg}")
```
